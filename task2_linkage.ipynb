```python
import pandas as pd
```

```python
census_df = pd.read_csv('data/MF_census_1850_1846_cl.csv.gz')
```

```python
# census_df.dropna(subset=['inp_PR_BIR_YEAR_EST', 'inp_PR_NAME'])
# census_df['inp_PR_SEX_CODE'] = census_df['inp_PR_SEX_CODE'].fillna('Unknown')

census_1846_df = census_df[census_df['inp_EVENT_YEAR'] == 1846]
census_1850_df = census_df[census_df['inp_EVENT_YEAR'] == 1850]
```

```python
linkage_columns = ['pa_id',
                   'inp_PR_BIR_YEAR_EST',
                   'inp_PR_SEX_CODE',
                   'inp_PR_NAME', 
                   'inp_PR_NAME_GN',
                   'inp_PR_NAME_SURN',
                   'inp_PR_BIR_PLACE_ORIG']

left = census_1846_df[linkage_columns]
right = census_1850_df[linkage_columns]

left = left.dropna(subset=['inp_PR_BIR_YEAR_EST', 'inp_PR_NAME'])
right = right.dropna(subset=['inp_PR_BIR_YEAR_EST', 'inp_PR_NAME'])

left['inp_PR_SEX_CODE'] = left['inp_PR_SEX_CODE'].fillna('Unknown')
right['inp_PR_SEX_CODE'] = right['inp_PR_SEX_CODE'].fillna('Unknown')

left['name_first_letter'] = left['inp_PR_NAME'].str[0].str.lower()
right['name_first_letter'] = right['inp_PR_NAME'].str[0].str.lower()


blocking_df = pd.merge(
    left,
    right,
    on = ['inp_PR_BIR_YEAR_EST','inp_PR_SEX_CODE', 'name_first_letter'],
    how = 'left',
    suffixes= ['_1846', '_1850']
)

# blocking_df.to_csv('output/blocking.csv')
```
